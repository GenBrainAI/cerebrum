## Run a test

Note: if run_pytest_in_notebook recieves a folder, all tests under folder will run

```python
# if you'd like to run a test
import pytest
import os

def run_pytest_in_notebook(test_path):
    # Run pytest and capture the output
    pytest_result = pytest.main([test_path])

    if pytest_result == 0:
        print("Tests passed successfully.")
    else:
        print("Some tests failed.")

os.chdir(base_path)
# Example of how to call it
run_pytest_in_notebook(base_path + '/tests/test_toolset.py')
```

## Load Unicorn server
Run once

```python
import uvicorn
from threading import Thread

def run_server():
    uvicorn.run("src.server.main:app", host="127.0.0.1", port=8000, log_level="info")

# Run the server in a separate thread
thread = Thread(target=run_server)
thread.start()
```

## Call an API

```python
import requests

# URL of the health_check endpoint
url = "http://127.0.0.1:8000/health_check"

# Make a GET request to the endpoint
response = requests.get(url)

# Check the response
if response.status_code == 200:
    print("All OK:", response.text)
else:
    print("Houston, we have a problem:", response.text)
```
